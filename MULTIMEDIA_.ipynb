```python
from PIL import Image
import google.generativeai as genai

# ✅ Configure Gemini
genai.configure(api_key="AIzaSyBHwYOOtaaLEfLU3WzaL-CW4UCDuZiTdbE")  # Replace with your real key
model = genai.GenerativeModel("models/gemini-1.5-flash")

# 🧠 Store personal info
memory = {}

def store_info(statement):
    lower_stmt = statement.lower()
    if "my name is" in lower_stmt:
        name = statement.split("my name is")[-1].strip().capitalize()
        memory["name"] = name
        return f"Nice to meet you, {name}!"
    elif "i am" in lower_stmt:
        name = statement.split("i am")[-1].strip().capitalize()
        memory["name"] = name
        return f"Nice to meet you, {name}!"
    elif "my favorite color is" in lower_stmt:
        color = statement.split("my favorite color is")[-1].strip()
        memory["favorite_color"] = color
        return f"Got it! Your favorite color is {color}."
    elif "i like" in lower_stmt:
        interest = statement.split("i like")[-1].strip()
        memory.setdefault("likes", []).append(interest)
        return f"Cool! I’ll remember you like {interest}."
    return None

def respond(question):
    if "what is my name" in question.lower() and "name" in memory:
        return f"Your name is {memory['name']}."
    if "what is my favorite color" in question.lower() and "favorite_color" in memory:
        return f"Your favorite color is {memory['favorite_color']}."
    if "what do i like" in question.lower() and "likes" in memory:
        return f"You like {', '.join(memory['likes'])}."

    # fallback to Gemini for generic questions
    response = model.generate_content(question)
    return response.text

# 🔁 Chat loop with memory
while True:
    question = input("You: ")
    if question.lower() in ["exit", "quit"]:
        print("Bot: Goodbye! 👋")
        break

    # Check if it's a personal info statement
    info_reply = store_info(question)
    if info_reply:
        print("Bot:", info_reply)
        continue

    # Generate answer with memory support
    answer = respond(question)
    print("Bot:", answer)
```
